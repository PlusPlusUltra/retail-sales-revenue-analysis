```python
"""# Retail Sales Revenue Analysis

## Project Objective

The objective of this project is to analyze e-commerce sales data to identify revenue trends, customer behavior, and profitability drivers.

## Dataset

This dataset contains Brazilian e-commerce transactions including orders, customers, products, and payments.

## Analysis goals:

• Clean datasets  
• Perform exploratory analysis  
• Identify KPIs  
• Generate business insights  """

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)

orders = pd.read_csv("../data/raw/olist_orders_dataset.csv")

customers = pd.read_csv("../data/raw/olist_customers_dataset.csv")

order_items = pd.read_csv("../data/raw/olist_order_items_dataset.csv")

products = pd.read_csv("../data/raw/olist_products_dataset.csv")

payments = pd.read_csv("../data/raw/olist_order_payments_dataset.csv")

orders.head()

orders.info()

orders.describe()
```

```python
orders.isnull().sum()

customers.isnull().sum()

order_items.isnull().sum()

products.isnull().sum()

payments.isnull().sum()
```

```python
orders.duplicated().sum()

customers.duplicated().sum()

order_items.duplicated().sum()

products.duplicated().sum()

payments.duplicated().sum()
```

```python
date_columns = [
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date'
]

for col in date_columns:
    orders[col] = pd.to_datetime(orders[col])
```

```python
######Missing delivery dates correspond primarily to orders that were not delivered. These values were retained because they reflect the true business process rather than data errors.
```

```python
orders['purchase_year'] = (
    orders['order_purchase_timestamp'].dt.year
)

orders['purchase_month'] = (
    orders['order_purchase_timestamp'].dt.month
)

orders['purchase_day'] = (
    orders['order_purchase_timestamp'].dt.day
)
```

```python
print(orders.shape)
print(customers.shape)
print(order_items.shape)
print(products.shape)
print(payments.shape)
```

```python
sales = orders.merge(
    customers,
    on='customer_id',
    how='left'
)
```

```python
sales = sales.merge(
    order_items,
    on='order_id',
    how='left'
)
```

```python
sales = sales.merge(
    products,
    on='product_id',
    how='left'
)
```

```python
#One order item purchased by one customer
```

```python
total_revenue = sales['price'].sum()

total_orders = sales['order_id'].nunique()

total_customers = sales['customer_unique_id'].nunique()

aov = total_revenue / total_orders
```

```python
sales.shape
```

```python
## Data Cleaning Decisions

### Missing Values

- Missing delivery dates were retained because they correspond to orders that were not completed or delivered.
- Missing product category information was retained for further investigation.

### Duplicates

- No fully duplicated rows were identified.

### Data Types

- Order date fields were converted from string format to datetime format.

### Feature Engineering

- Created purchase_year
- Created purchase_month
- Created purchase_day
```
